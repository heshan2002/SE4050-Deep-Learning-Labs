## Task 1.2
Create a random 100000×1 array from an exponential distribution and visualize it together with uniform and normal distributions.


```python
sample_size = 100000

exponential_data = np.random.exponential(scale=1.0, size=sample_size)
uniform_data = np.random.uniform(0, 1, size=sample_size)
normal_data = np.random.normal(0, 1, size=sample_size)

plt.figure(figsize=(10, 6))

plt.hist(exponential_data, density=True, bins=120, histtype="step",
         linewidth=1.5, label="Exponential")

plt.hist(uniform_data, density=True, bins=120, histtype="step",
         linewidth=1.5, label="Uniform")

plt.hist(normal_data, density=True, bins=120, histtype="step",
         linewidth=1.5, label="Normal")

plt.xlim(-4, 8)
plt.title("Random Distributions")
plt.xlabel("Value")
plt.ylabel("Density")
plt.legend(loc="upper right")
plt.grid(alpha=0.25)

plt.show()
```

## Task 1.3
Using Matplotlib, plot:

**Z = X² + Y²**

Limit X and Y values to **[-5, 5]**.


```python
from mpl_toolkits.mplot3d import Axes3D

x = np.linspace(-5, 5, 100)
y = np.linspace(-5, 5, 100)

X, Y = np.meshgrid(x, y)
Z = X**2 + Y**2

fig = plt.figure(figsize=(10, 7))
ax = fig.add_subplot(111, projection="3d")

surface = ax.plot_surface(X, Y, Z, cmap="viridis")

ax.set_title("3D Plot of Z = X² + Y²")
ax.set_xlabel("X")
ax.set_ylabel("Y")
ax.set_zlabel("Z")

fig.colorbar(surface, shrink=0.6, aspect=12)

plt.show()
```

## Task 1.4
Calculate Pearson correlation and Spearman rank correlation for:

- HP
- Attack
- Defense
- Sp. Atk
- Sp. Def
- Speed

Then visualize the results using heatmaps.

### Dataset
Upload the dataset used with `seaborn_tutorial.ipynb`.

The code below allows you to upload the required CSV file directly in Google Colab.


```python
from google.colab import files
uploaded = files.upload()
```

```python
# Replace the filename below with the uploaded CSV filename if needed.
# The code automatically uses the first uploaded file.

uploaded_filename = list(uploaded.keys())[0]
pokemon_df = pd.read_csv(uploaded_filename)

print("Loaded file:", uploaded_filename)
display(pokemon_df.head())
```

```python
# Show column names to confirm exact spelling
print(pokemon_df.columns.tolist())
```

```python
features = ['HP', 'Attack', 'Defense', 'Sp. Atk', 'Sp. Def', 'Speed']

selected_df = pokemon_df[features]

print("Selected features:")
display(selected_df.head())
```

### Pearson Correlation


```python
pearson_corr = selected_df.corr(method='pearson')

print("Pearson Correlation Matrix:")
display(pearson_corr)

plt.figure(figsize=(9, 7))
sns.heatmap(
    pearson_corr,
    annot=True,
    fmt=".2f",
    cmap="coolwarm",
    square=True,
    linewidths=0.5
)

plt.title("Pearson Correlation Heatmap")
plt.show()
```

### Spearman Rank Correlation


```python
spearman_corr = selected_df.corr(method='spearman')

print("Spearman Rank Correlation Matrix:")
display(spearman_corr)

plt.figure(figsize=(9, 7))
sns.heatmap(
    spearman_corr,
    annot=True,
    fmt=".2f",
    cmap="coolwarm",
    square=True,
    linewidths=0.5
)

plt.title("Spearman Rank Correlation Heatmap")
plt.show()
```

## Task 2.4
Plot the histogram of each feature in a **1×4 layout**.


```python
features = ['N_total', 'N_bulk', 'N_surface', 'R_avg']

fig, axes = plt.subplots(1, 4, figsize=(20, 5))

for i, feature in enumerate(features):
    axes[i].hist(new_df[feature], bins=30, edgecolor='black')
    axes[i].set_title(feature)
    axes[i].set_xlabel(feature)
    axes[i].set_ylabel("Frequency")

plt.tight_layout()
plt.show()
```

## Task 2.5
Visualize scatter plots and histograms using Seaborn `pairplot`.


```python
sns.pairplot(new_df)
plt.show()
```

## Task 2.6
Use `PairGrid` so that:

- Upper half = bivariate histogram
- Diagonal = histogram + KDE
- Lower half = bivariate KDE


```python
g = sns.PairGrid(new_df)

g.map_upper(sns.histplot)
g.map_diag(sns.histplot, kde=True)
g.map_lower(sns.kdeplot, fill=True)

plt.show()
```
